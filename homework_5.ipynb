### W5A16: Viết hàm kiểm tra một số có phải số Armstrong không?

```python
def amstrong(x):
    y = str(x)
    t = len(y)
    s = 0
    for d in y:
        s += int(d)**t 
    if s == x:
        return True
    return False
x = int(input())
amstrong(x)
```

### W5A18: Kiểm tra số đối xứng

```python
def so_doi_xung(x):
    t = 0
    s = x
    while s > 0:
        t = t * 10 + s % 10
        s //= 10
        if t == x:
            return True
    return False
x = int(input())
so_doi_xung(x)    
```

### W5A19: Kiểm tra số kỳ diệu (tổng chữ số lặp về 1)

```python
def so_ki_dieu(n):
    while n > 9:
        s = 0
        while n > 0:
          s += n % 10
          n //= 10
        n = s    
    return n == 1
n = int(input())
so_ki_dieu(n)     
```

### W5A24: iết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.

```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    return (y2 - y1) * (x3 - x1) == (x2 - x1) * (y3 - y1)
x1, y1, x2, y2, x3, y3 = map(int, input().split())
are_collinear(x1, y1, x2, y2, x3, y3)
```

### W5A25: Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay không.

```python
def tich_co_huong(A, B, P):
    return (B[0] - A[0]) * (P[1] - A[1]) - (B[1] - A[1]) * (P[0] - A[0])
def point_in_quadrilateral(quad, p):
    A, B, C, D = quad
    c1 = tich_co_huong(A, B, p)
    c2 = tich_co_huong(B, C, p)
    c3 = tich_co_huong(C, D, p)
    c4 = tich_co_huong(D, A, p)
    if (c1 >= 0 and c2 >= 0 and c3 >= 0 and c4 >= 0) or (c1 <= 0 and c2 <= 0 and c3 <= 0 and c4 <= 0):
        return True
    return False
quad = []
for i in range(4):
    x, y = map(int, input().split())
    quad.append((x, y))
px, py = map(int, input().split())
p = (px, py)
point_in_quadrilateral(quad, p)    
```

### W5A27: Đếm chữ lượng chữ sô 0 trong một list số nguyên

```python
def count_zero(numbers):
    count = 0
    for i in numbers:
        for d in str(abs(i)):
            if d != '0':
                count += 1
    return count
numbers = list(map(int, input().split()))
count_zero(numbers)            
```

### W5A28: Tính góc ABC từ tọa độ A(x1,y1) B(x2,y2) và C(x3,y3)

```python
import math
def tinh_goc(A, B, C):
    BA = (A[0] - B[0], A[1] - B[1])
    BC = (C[0] - B[0], C[1] - B[1])
    tu = BA[0] * BC[0] + BA[1] * BC[1]
    mau1 = math.sqrt(BA[0] ** 2 + BA[1] ** 2)
    mau2 = math.sqrt(BC[0] ** 2 + BC[1] ** 2)
    cos_ABC = tu / (mau1 * mau2)
    return math.degrees(math.acos(cos_ABC))
x1, y1, x2, y2, x3, y3 = map(int, input().split())
A = (x1, y1)
B = (x2, y2)
C = (x3, y3)
tinh_goc(A, B, C)
```

### W5A30: Viết hàm tính hệ số nhị thức newton C(n,k)

```python
def C(n,k):
    if k > n:
        return 0
    res = 1
    for i in range(1, k + 1):
        res = res * ( n - i + 1 ) // i
    return res
n, k = map(int, input().split())
C(n,k)    
```

### W5A33: tính tiền taxi 

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    fare = 0
    

    if car_type == 1:
        fare += 10000
    else:
        fare += 15000
    
    fare += km * 12000
    fare += wait_minutes * 500
    

    if start_hour >= 22 or start_hour < 5:
        fare *= 1.2
    
    return fare
```

### W5A34: Viết hàm trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau

```python
def xet_x(x):
    s = str(x)
    return len(set(s)) == len(s)
def sum_1_to_n(n):
    if n <= 0:
        return 0
    total = 0
    for i in range(1, n + 1):
        if xet_x(i):
            total += i
    return total
n = int(input())
sum_1_to_n(n)        
```
