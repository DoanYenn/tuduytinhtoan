### W5A16: Viết hàm kiểm tra một số có phải số Armstrong không?

```python
def is_armstrong(n):
    num_str = str(n)
    num_digits = len(num_str)
    
    total = 0
    for digit_char in num_str:
        digit = int(digit_char)
        total += digit ** num_digits
    
    return total == n

n = int(input("Nhập số cần kiểm tra: "))
if is_armstrong(n):
    print(f"{n} là số Armstrong")
else:
    print(f"{n} không là số Armstrong")
```

### W5A17: Kiểm tra số chính phương

```python
def is_perfect_square(n):
    if n < 0:
        return False
    i = 1
    while i * i <= n:
        if i * i == n:
            return True
        i += 1
    return False

n = int(input("Nhập số: "))
print(is_perfect_square(n))
```

### W5A18: Kiểm tra số đối xứng

```python
def is_palindrome(n):
    if n < 0:
        return False
    
    original = n
    reversed_num = 0
    
    while n > 0:
        reversed_num = reversed_num * 10 + n % 10
        n //= 10
    
    return original == reversed_num

n = int(input("Nhập số: "))
print(is_palindrome(n))
```

### W5A19: Kiểm tra số kỳ diệu (tổng chữ số lặp về 1)

```python
def is_magic_number(n):

    while n >= 10:
        total = 0
        while n > 0:
            total += n % 10
            n //= 10
        n = total
    return n == 1

n = int(input("Nhập số: "))
if is_magic_number(n):
    print(f"{n} là số kỳ diệu")
else:
    print(f"{n} không là số kỳ diệu")
```

### W5A20: Viết hàm detect_number_type(s) trả về 'int', 'float', hoặc 'neither'.

```python
def detect_number_type(s):
    i = 0
    if s[0] == '-':
        i = 1

    has_dot = False

    while i < len(s):
        if s[i] == '.':
            if has_dot:
                return 'neither'
            has_dot = True
        elif s[i] < '0' or s[i] > '9':
            return 'neither'
        i += 1

    if has_dot:
        return 'float'
    return 'int'
```

### W5A21: Viết hàm kiểm tra một trong ba số bằng tích hai số còn lại.

```python
def is_one_product_of_others(a, b, c):
    if a == b * c:
        return True
    if b == a * c:
        return True
    if c == a * b:
        return True
    return False
a, b, c = map(int, input().split())
is_one_product_of_others(a, b, c)
```

### W5A22: Viết hàm nhận chuỗi 6 chữ số và trả về tổng 3 chữ số đầu + 3 chữ số cuối.

```python
def sum_halves(s):
    if len(s) != 6 or not s.isdigit():
        return 0
    first = int(s[:3])
    second = int(s[3:])
    return first + second
s = input()
sum_halves(s)
```

### W5A23: Viết hàm quadratic_root_count(a, b, c) trả về số nghiệm thực của phương trình a* x^2 + b*x + c = 0.

```python
def quadratic_roots_count(a, b, c):
    d = b * b - 4 * a * c

    if d > 0:
        return 2
    elif d == 0:
        return 1
    else:
        return 0
a, b, c = map(int, input().split())
quadratic_roots_count(a, b, c)
```

### W5A24: iết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.

```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    return (y2 - y1) * (x3 - x2) == (y3 - y2) * (x2 - x1)
x1, y1, x2, y2, x3, y3 = map(int, input().split())
are_collinear(x1, y1, x2, y2, x3, y3)
```
