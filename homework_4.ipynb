### W4A20: Kiểm tra n có dạng 2^k không

```python
n = int(input())
f = False
for i in range(1, n):
    if i * i == n:
        f = True
        break
    else:
        f = False
if f:
    print("n co dang 2^k")
else:
    print("n khong co dang 2^k")            
```

### W4A21: Tính tổng các chữ số của n

```python
n = int(input())
tong = 0
while n > 0:
    tong += n % 10
    n //= 10
print(tong)     
```

### W4A22: Đếm số chữ số chẵn và lẻ

```python
n = int(input())
chan = 0
le = 0
while n > 0:
    digit = n % 10
    if digit % 2 == 0:
        chan += 1
    else:
        le += 1
    n //= 10
print("chu co chan: ", chan)
print("chu so le: ", le)            
```

### W4A24: Nhập vào A. Tìm n nhỏ nhất sao cho 1 + 1/2 + 1/3 + ... + 1/n > A

```python
A = int(input())
s = 0
n = 0
while (s + 1/(n + 1)) <= A:
    n += 1
    s += 1/n
print(f" n nho nhat la {n} ")     
```

### W4A25: Nhập dãy số, dừng khi nhập -1. In số lớn nhất và nhỏ nhất

```python
max_val = None
min_val = None
while True:
    x = int(input(" nhap so ( nhap -1 de dung )"))
    if x == -1:
        break
    if max_val is None or x > max_val:
        max_val = x
    if min_val is None or x < min_val:
        min_val = x    
print("Max:", max_val)
print("Min:", min_val)        
```

### W4A26: Nhập vào số nguyên dương A. Tìm số Fibonacci lớn nhất không vượt quá A

```python
A = int(input())
a, b = 1, 1
while b <= A:
    a, b = b, a + b
print(a)     
```

### W4A27: Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu từ.(Quy định: chỉ gồm chữ cái và khoảng trắng, không có ký tự đặc biệt)

```python
s = input("nhap chuoi")
count = 0
i = 0
while True:
    try:
        s[i]
    except:
        break
    if s[i] != ' ' and (i == 0 or s[i - 1] == ' '):
        count += 1
    i += 1
print("so tu ", count)            
```

### W4A28: Nhập vào một chuỗi, hãy in từ đầu tiên trong chuỗi

```python
s = input()
word = ""
i = 0
while True:
    try:
        s[i]
    except:
        break
    if s[i] == ' ':
        break
    word += s[i]
    i += 1
print(word)         
```

### W4A29: Cho một chuỗi chỉ gồm chữ hoa và chữ thường, hãy đổi chữ cái thường thành hoa và hcuwx cái hoa thành thường

```python
s = input()
word = ""
for i in s:
    if 'a' <= i <= 'z':
        word += chr(ord(i) - 32)
    elif 'A' <= i <= 'Z':
        word += chr(ord(i) + 32)
    else:
        word += i
print(word)                
```

### W4A30: nhập vào một chuỗi dạng 3 số nguyên, mỗi số nguyên cách nhau môt dấu phẩy, hãy tính tổng 3 số đó
