W6A16: Nhập vào mảng n hàng, m cột số nguyên.
In lại mảng ra màn hình như ma trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải.

```python
n, m = map(int, input().split())

a = []

for _ in range(n):
    row = list(map(int, input().split()))
    a.append(row)

for i in range(n):
    for j in range(m):
        print(f"{a[i][j]:>4}", end="")
    print()
```

W6A17: Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và đường chéo phụ của mảng.

```python
n = int(input())

a = []

for _ in range(n):
    row = list(map(int, input().split()))
    a.append(row)

# Đường chéo chính
for i in range(n):
    print(a[i][i], end=" ")

print()

# Đường chéo phụ
for i in range(n):
    print(a[i][n - 1 - i], end=" ")
```

W6A18: Nhập vào mảng n hàng, m cột số nguyên và số nguyên k

Tính và in ra tổng các phần tử trên cột k (đánh số từ 0).

W6A19: Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất trong list.

```python
def max_pos(nums):
    pos = 0

    for i in range(1, len(nums)):
        if nums[i] > nums[pos]:
            pos = i

    return pos


a = list(map(int, input().split()))

print(max_pos(a))
```

W6A20: Viết hàm đưa vào một list số nguyên và một số nguyên dương k.
Hãy tìm và trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về -1.

```python
def find_first(nums, k):
    for i in range(len(nums)):
        if nums[i] == k:
            return i

    return -1


a = list(map(int, input().split()))
k = int(input())

print(find_first(a, k))
```

W6A21: Nhập một dòng gồm các chuỗi (tách bởi dấu cách).
Tạo dictionary ánh xạ:

```python
words = input().split()

d = {}

for i in range(len(words)):
    w = words[i]

    if w not in d:
        d[w] = ()

    d[w] += (i,)

print(d)
```

W6A23: Viết hàm xóa đi các số < x có trong mảng.

```python
def remove_smaller(nums, x):
    res = []

    for v in nums:
        if v >= x:
            res.append(v)

    return res


a = list(map(int, input().split()))
x = int(input())

print(remove_smaller(a, x))
```

W6A24: Viết hàm đẩy tất cả các số 0 về đầu (hoặc cuối), các số khác vẫn đúng vị trí tương đối trước sau.

```python
def move_zero_front(nums):
    zeros = []
    others = []

    for x in nums:
        if x == 0:
            zeros.append(x)
        else:
            others.append(x)

    return zeros + others


a = list(map(int, input().split()))

print(move_zero_front(a))q
```
