### W5A1: Viết hàm nhận vào hai số nguyên a, b. Trả về số lớn hơn.

```python
def Max_Of_Two(a, b):
    if a > b:
        return a
    return b
a, b = map(int, input().split())
M = Max_Of_Two(a,b)
print(M)
```

### W5A3: Kiểm tra một số có phải số nguyên tố không.

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
n = int(input())
is_prime(n)
```

### W5A4: Kiểm tra số hoàn hảo.

```python
def is_perfect(n):
    if n < 1:
        return False
    total = 0
    for i in range(1, n):
        if n % i == 0:
            total += i
    return total == n
n = int(input())
is_perfect(n)
```

### W5A5: Tìm vị trí phần tử đầu tiên = k trong list (không có trả -1)

```python
def find_first(lst, k):
    for i in range(len(lst)):
        if lst[i] == k:
            return i + 1
    return -1
lst = list(map(int, input().split()))
k = int(input())
find_first(lst, k)
```

### W5A6: Tính giai thừa n!

```python
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
n = int(input())
factorial(n)
```

### W5A8: Khoảng cách Hamming giữa 2 số nguyên

```python
def hamming_distance(x, y):
    count = 0
    while x > 0 or y > 0:
        if (x % 2) != (y % 2):
            count += 1
        x //= 2
        y //= 2
    return count
x = int(input())
y = int(input())
hamming_distance(x, y)
```

### W5A9: Tổng chữ số của số nguyên dương n

```python
def sum_digits(n):
    total = 0
    while n > 0:
        total += n % 10
        n //= 10
    return total
n = int(input())
sum_digits(n)
```

### W5A10: Kiểm tra 2 từ có phải dạng cấu (isomorphic)

```python
def is_isomorphic(a, b):
    if len(a) != len(b):
        return False
    
    map_ab = {}
    map_ba = {}
    
    for i in range(len(a)):
        if a[i] in map_ab:
            if map_ab[a[i]] != b[i]:
                return False
        else:
            map_ab[a[i]] = b[i]
        
        if b[i] in map_ba:
            if map_ba[b[i]] != a[i]:
                return False
        else:
            map_ba[b[i]] = a[i]
    
    return True
```

### W5A12: Viết hàm divide_with_remainder(dividend, divisor) trả về bộ (q, r) sao cho dividend = divisor*q + r và 0 <= |r| < |divisor|.

```python
def divide_with_remainder(dividend, divisor):
    q = dividend // divisor
    r = dividend % divisor
    return q, r
dividend = int(input())
divisior = int(input())
divide_with_remainder(dividend, divisior)
```

### W5A15: Viết hàm is_right_triangle(a, b, c) trả về True nếu ba số có thể là ba cạnh của tam giác vuông.

```python
def is_right_triangle(a, b, c):
    sides = sorted([a, b, c])
    return sides[0]**2 + sides[1]**2 == sides[2]**2
a, b, c = map(int,input().split())
is_right_triangle(a, b, c)
```
