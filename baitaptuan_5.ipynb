### W5A1: Viết hàm nhận vào hai số nguyên a, b. Trả về số lớn hơn.

```python
def max_a_b(a, b):
    if a > b:
        return a
    if b > a:
        return b
a, b = map(int, input().split())
max_a_b(a ,b)    
```

### W5A3: Kiểm tra một số có phải số nguyên tố không.

```python
def so_nguyen_to(a):
    if a < 2:
        return False
    else:
        for i in range(2, a):
            if a % i == 0:
                return False
    return True
a = int(input())
so_nguyen_to(a)        
```

### W5A4: Kiểm tra số hoàn hảo.

```python
def perfect_number(n):
    tong = 0
    for i in range(1, n):
        if n % i == 0:
            tong += i
    if tong == n:
        return True
    return False
n = int(input())
perfect_number(n)        
```

### W5A5: Tìm vị trí phần tử đầu tiên = k trong list (không có trả -1)

```python
def find_k(number,k):
    for i in range(0, len(number)):
        if number[i] == k:
            return i + 1
    return -1
number = list(map(int, input().split()))
k = int(input())
find_k(number, k)    
```

### W5A6: Tính giai thừa n!

### W5A7: Máy tính đơn giản (+, -, *, /), làm tròn 2 chữ số

```python
def may_tinh(a, op, b):
    if op == '+':
        return a + b
    if op == '-':
        return a - b
    if op == '*':
        return a * b
    if op == '/':
        return a / b
a = float(input())
b = float(input())
op = input()
s = may_tinh(a, op, b)
print(f"{s:.2f}")
```

### W5A8: Khoảng cách Hamming giữa 2 số nguyên

```python
def hamming(n):
    s = ""
    while n:
        s = chr((n % 2) + 48) + s
        n //= 2
    return s
a = int(input())
b = int(input())
day_a = hamming(a)
day_b = hamming(b)
while len(day_a) < len(day_b):
    day_a = "0" + day_a
while len(day_a) > len(day_b):
    day_b = "0" + day_b
count = 0
for i in range(len(day_a)):
    if day_a[i] != day_b[i]:
        count += i
print(count)                    
```

### W5A9: Tổng chữ số của số nguyên dương n

```python
def tong_chu_so(n):
    t = 0
    while n > 0:
        t = t + n % 10
        n //= 10
    return t
n = int(input())
tong_chu_so(n)    
```

### W5A10: Kiểm tra 2 từ có phải dạng cấu (isomorphic)

```python
def dang_cau(s, t):
    if len(s) != len(t):
        return False
    map_s_t = {}
    map_t_s = {}
    for i in range(len(s)):
        c1 = s[i]
        c2 = t[i]
    if c1 in map_s_t:
        if map_s_t[c1] != c2:
            return False
    else:
        map_s_t[c1] = c2
    if c2 in map_t_s:
        if map_t_s[c2] != c1:
            return False
    else:
        map_t_s[c2] = c1        
    return True
s = input()
t = input()
dang_cau(s, t)
```

### W5A12: Viết hàm divide_with_remainder(dividend, divisor) trả về bộ (q, r) sao cho dividend = divisor*q + r và 0 <= |r| < |divisor|.

```python
def divide_with_remainder(dividend, divior):
    q = dividend // divior
    r = dividend - divior*q
    return q, r
dividend = int(input())
divior = int(input())
divide_with_remainder(dividend, divior)
```

### W5A14: Viết hàm compute(a, b, op) thực hiện phép tính a (+|-|*|/) b tùy theo ký tự op.

```python
def calculator(num1, num2, op):
    if op == '+':
        result = num1 + num2
    elif op == '-':
        result = num1 - num2
    elif op == '*':
        result = num1 * num2
    elif op == '/':
        result = num1 / num2
    else:
        return None
    return round(result, 2)
a = float(input())
op = input()
b = float(input())
calculator(a, b ,op)
```

### W5A15: Viết hàm is_right_triangle(a, b, c) trả về True nếu ba số có thể là ba cạnh của tam giác vuông.

```python
def is_right_triangle(a, b, c):
    if a ** 2 == b ** 2 + c**2 or b ** 2 == a ** 2 + c ** 2 or c**2 == a**2 + b**2:
        return True
    return False
a, b, c = map(int, input().split())
is_right_triangle(a, b, c) 
```
