Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu mỗi phần tử x trong mảng xuất hiện đúng x lần.

Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa đi y số (y nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.

```python
a = list(map(int, input().split()))

dem = {}
for x in a:
    dem[x] = dem.get(x, 0) + 1

xoa = 0

for x in dem:
    if dem[x] < x:
        xoa += dem[x]
    elif dem[x] > x:
        xoa += dem[x] - x

print(xoa)
```

Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng bằng thuật toán Boyer-Moore.

```python
a = list(map(int, input().split()))

ung_vien = None
dem = 0

for x in a:
    if dem == 0:
        ung_vien = x
        dem = 1
    elif x == ung_vien:
        dem += 1
    else:
        dem -= 1

if a.count(ung_vien) > len(a) // 2:
    print(ung_vien)
```

Số chẵn đứng trước, số lẻ đứng sau; chẵn tăng dần, lẻ giảm dần.

```python
a = list(map(int, input().split()))

chan = []
le = []

for x in a:
    if x % 2 == 0:
        chan.append(x)
    else:
        le.append(x)

chan.sort()
le.sort(reverse=True)

print(*(chan + le))
```

Merge hai mảng đã sắp xếp (tăng hoặc giảm) mà không dùng thuật toán sắp xếp.

```python
a = list(map(int, input().split()))
b = list(map(int, input().split()))

if len(a) > 1 and a[0] > a[-1]:
    a.reverse()

if len(b) > 1 and b[0] > b[-1]:
    b.reverse()

i = j = 0
c = []

while i < len(a) and j < len(b):
    if a[i] <= b[j]:
        c.append(a[i])
        i += 1
    else:
        c.append(b[j])
        j += 1

while i < len(a):
    c.append(a[i])
    i += 1

while j < len(b):
    c.append(b[j])
    j += 1

print(*c)
```

Tên sinh viên xuất hiện nhiều nhất

```python
n = int(input())

dem = {}

for _ in range(n):
    ten_day_du = input().strip()
    ten = ten_day_du.split()[-1]

    dem[ten] = dem.get(ten, 0) + 1

max_dem = max(dem.values())

kq = []

for ten in dem:
    if dem[ten] == max_dem:
        kq.append(ten)

print(min(kq))
```

Số lớn nhất có đúng k chữ số

```python
s = input()
k = int(input())

stack = []
xoa = len(s) - k

for c in s:
    while stack and xoa > 0 and stack[-1] < c:
        stack.pop()
        xoa -= 1
    stack.append(c)

while xoa > 0:
    stack.pop()
    xoa -= 1

print(''.join(stack[:k]))
```

Tách số trong văn bản

```python
import re

s = input()

ds = re.findall(r'\d+', s)

nums = [int(x) for x in ds]

tong = sum(nums)

if len(nums) > 0:
    tb = tong / len(nums)
else:
    tb = 0

print("Tong =", tong)
print("Trung binh =", tb)
```

Kiểm tra dãy ngoặc hợp lệ

```python
s = input()

stack = []

dong = {
    ')': '(',
    ']': '[',
    '}': '{'
}

hop_le = True

for c in s:
    if c in "([{":
        stack.append(c)
    else:
        if not stack or stack[-1] != dong[c]:
            hop_le = False
            break
        stack.pop()

if stack:
    hop_le = False

print(hop_le)
```

Dãy hợp số liên tiếp dài nhất

```python
def hop_so(n):
    if n < 4:
        return False

    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return True
    return False

N = int(input())

best_l = best_r = 0
i = N + 1

while i <= N + 1000:
    if hop_so(i):
        l = i

        while i <= N + 1000 and hop_so(i):
            i += 1

        r = i - 1

        if r - l > best_r - best_l:
            best_l = l
            best_r = r
    else:
        i += 1

print(best_l, best_r)
```

Dãy con tăng liên tục dài nhất

```python
a = list(map(int, input().split()))

best = []
cur = [a[0]]

for i in range(1, len(a)):
    if a[i] > a[i-1]:
        cur.append(a[i])
    else:
        if len(cur) > len(best):
            best = cur[:]
        cur = [a[i]]

if len(cur) > len(best):
    best = cur

print(*best)
```

Dãy con tăng không liên tục dài nhất (LIS)

```python
a = list(map(int, input().split()))

n = len(a)

dp = [1] * n

for i in range(n):
    for j in range(i):
        if a[j] < a[i]:
            dp[i] = max(dp[i], dp[j] + 1)

print(max(dp))
```
