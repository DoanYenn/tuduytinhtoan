### W4A7: Nhập vào số nguyên dương n (n ≥ 2). Hãy in ra ước số nguyên tố lớn nhất của n.

### W4A8:Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số đối xứng (palindrome). In ra số bước và giá trị palindrome thu được.

```python
def reverse(n):
    rev = 0
    while n > 0:
        rev = rev * 10 + n % 10
        n //= 10
    return(rev)
n = int(input())
step = 0
while n != reverse(n):
    n = n + reverse(n)
    step += 1
print(n, step)    
```

### W4A9: Nhập vào số nguyên dương n. In ra tất cả các số chính phương ≤ n có các chữ số đôi một khác nhau (in theo thứ tự tăng dần, cùng hàng, cách nhau bởi một dấu cách). Nếu không có, in ra (no number).

```python
def chinh_phuong(n):
    while n > 0:
        d = n % 10
        t = n // 10
        while t > 0:
            if d == t % 10:
                return False
            t //= 10
        n //= 10
    return True    
n = int(input())
found = False
i = 1
while i * i <= n:
    k = i * i
    if chinh_phuong(k):
        print(k ,end=" ")
        found = True
    i += 1
if not found:
    print("no number")        


```

### W4A13

```python
def uoc(n):
    s = 0   
    for i in range(1, n):
        if n % i == 0:
            s += i
    return s
def kiem_tra_than_thiet(a,b):
    tong_uoc_a = uoc(a)
    tong_uoc_b = uoc(b)
    if tong_uoc_a == b and tong_uoc_b == a and a != b:
        return True
    else:
        return False
a = int(input())
b = int(input())
if kiem_tra_than_thiet(a, b):
    print("true")
else:
    print("false")    
```
