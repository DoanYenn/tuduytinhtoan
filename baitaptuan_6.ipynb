### W6A1: Nhập vào một dòng gồm các số nguyên. Hãy loại bỏ phần tử trùng nhưng giữ nguyên thứ tự xuất hiện đầu tiên, rồi in ra list kết quả. Yêu cầu in ra đúng dạng list Python

```python
s = input().split()
nums = list(map(int, s))
re = []
for x in nums:
    if x not in re:
        re.append(x)
print(re)
```

### W6A4: Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo dictionary ánh xạ mỗi key → list các value theo thứ tự xuất hiện. Nếu key lặp lại thì thêm vào list.

```python
pair = input().split()
d = {}
for p in pair:
    key, value = p.split(':')
    if key not in d:
        d[key] = []
    d[key].append(value)
print(d)        
```

### W6A5: Nhập một list số nguyên. In ra dictionary:

```python
nums = list(map(int, input().split()))

d = {'positives': 0, 'negatives': 0, 'zeros': 0}

for x in nums:
    if x > 0:
        d['positives'] += 1
    elif x < 0:
        d['negatives'] += 1
    else:
        d['zeros'] += 1

print(d)
```

### W6A9: Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một dictionary mới trong đó:
### Nếu một key xuất hiện trong cả hai dict → cộng giá trị
### Nếu chỉ có trong một dict → giữ nguyên
### In ra dictionary mới theo thứ tự từ điển tăng dần của key.

```python
d1 = eval(input())
d2 = eval(input())
res = {} 
for k in d1:
    res[k] = d1[k]
for k in d2:
    if k in res:
        res[k] += d2[k]
    else:
        res[k] = d2[k]
res = dict(sorted(res.items))
print(res)                
```

### W6A10: Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên dương k. In ra list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là một tuple. Thứ tự xuất hiện: chỉ lấy (i < j). Có thể trùng giá trị nhưng khác vị tríSắp xếp theo phần tử đầu tăng dần

```python
nums = list(map(int, input().split()))
k = int(input())
res = []
for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        if nums[i] + nums[j] == k:
            res.append((nums[i], nums[j]))
res.sort()
print(res)            
```

### W6A11: Nhập vào một tuple chứa các số nguyên. Tạo ra hai tuple mới: một chứa các số chẵn một chứa các số lẻ Giữ nguyên thứ tự tương đối. In ra mỗi tuple trên một dòng.

### W6A12: Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong list. Nếu có nhiều số cùng số lần xuất hiện lớn nhất → in ra số nhỏ nhất trong số đó.

```python
nums = list(map(int, input().split()))

d = {}

# đếm số lần xuất hiện
for x in nums:
    if x in d:
        d[x] += 1
    else:
        d[x] = 1

# tìm số lần xuất hiện lớn nhất
max_count = 0
for v in d.values():
    if v > max_count:
        max_count = v

# tìm số nhỏ nhất có tần suất = max_count
res = None
for k in d:
    if d[k] == max_count:
        if res is None or k < res:
            res = k

print(res)
```

### W6A13: Nhập vào một dictionary có value là duy nhất (không trùng nhau). Hãy tạo một dictionary mới bằng cách đảo ngược key và value của dictionary ban đầu. In ra dictionary mới.
