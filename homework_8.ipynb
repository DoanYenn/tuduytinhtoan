Xây dựng lớp Shape với phương thức area().
Xây dựng lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape, ghi đè area().
Lấy π = 3.14

```python
class Shape:
    def area(self):
        return 0


class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def area(self):
        return self.w * self.h


class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14 * self.r * self.r


kind = input()

if kind == "RECTANGLE":
    w, h = map(float, input().split())
    obj = Rectangle(w, h)
else:
    r = float(input())
    obj = Circle(r)

print(f"{obj.area():.2f}")
```

W8A17. Vehicle - Car

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def info(self):
        return f"{self.brand}-{self.speed}"


class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        super().__init__(brand, speed)
        self.seats = seats

    def info(self):
        return f"{self.brand}-{self.speed}-{self.seats}"


kind = input()

if kind == "Vehicle":
    brand, speed = input().split()
    obj = Vehicle(brand, speed)
else:
    brand, speed, seats = input().split()
    obj = Car(brand, speed, seats)

print(obj.info())
```

W8A18. Polygon - Triangle
Xây dựng lớp Polygon với phương thức perimeter().
Triangle(a,b,c) kế thừa Polygon và trả về a+b+c.

```python
class Polygon:
    def perimeter(self):
        return 0


class Triangle(Polygon):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def perimeter(self):
        return self.a + self.b + self.c


a, b, c = map(int, input().split())

t = Triangle(a, b, c)

print(t.perimeter())
```

A20. Account - SavingAccount
Cộng lãi: balance * rate vào số dư.

```python
class Account:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, x):
        self.balance += x

    def get_balance(self):
        return self.balance


class SavingAccount(Account):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate

    def add_interest(self):
        self.balance += self.balance * self.rate


balance, rate = map(float, input().split())

acc = SavingAccount(balance, rate)

acc.add_interest()

print(f"{acc.get_balance():.2f}")
```

W8A22. Matrix2x2
trace = a + d
determinant = ad - bc

```python
class Matrix2x2:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d

    def trace(self):
        return self.a + self.d

    def determinant(self):
        return self.a * self.d - self.b * self.c


a, b, c, d = map(int, input().split())

m = Matrix2x2(a, b, c, d)

print(m.trace(), m.determinant())
```

W8A23. Polynomial bậc nhất
P(x) = ax + b

```python
class Polynomial:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def evaluate(self, x):
        return self.a * x + self.b


a, b, x = map(int, input().split())

p = Polynomial(a, b)

print(p.evaluate(x))
```

Polynomial bậc hai
P(x)=ax²+bx+c

```python
class Polynomial2:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def evaluate(self, x):
        return self.a * x * x + self.b * x + self.c


a, b, c, x = map(int, input().split())

p = Polynomial2(a, b, c)

print(p.evaluate(x))
```

W8A25. Tổng hai đa thức
P(x)=ax+b
Q(x)=cx+d
In dạng kx + m.

```python
class PolynomialSum:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d

    def sum_str(self):
        k = self.a + self.c
        m = self.b + self.d
        return f"{k}x + {m}"


a, b, c, d = map(int, input().split())

p = PolynomialSum(a, b, c, d)

print(p.sum_str())
```

W8A27. DiscountProduct

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def final_price(self):
        return self.price


class DiscountProduct(Product):
    def __init__(self, name, price, percent):
        super().__init__(name, price)
        self.percent = percent

    def final_price(self):
        return self.price * (100 - self.percent) / 100


name, price, percent = input().split()

obj = DiscountProduct(name, float(price), float(percent))

print(f"{obj.final_price():.2f}")
```

W8A28. Counter

```python
class Counter:
    def __init__(self):
        self.count = 0

    def inc(self):
        self.count += 1

    def dec(self):
        self.count -= 1

    def value(self):
        return self.count


n = int(input())

ops = input().split()

c = Counter()

for op in ops:
    if op == "INC":
        c.inc()
    else:
        c.dec()

print(c.value())
```

W8A29. Time

```python
class Time:
    def __init__(self, h, m, s):
        self.h = h
        self.m = m
        self.s = s

    def to_seconds(self):
        return self.h * 3600 + self.m * 60 + self.s


h, m, s = map(int, input().split())

t = Time(h, m, s)

print(t.to_seconds())
```

W8A30. Stack

```python
class Stack:
    def __init__(self):
        self.data = []

    def push(self, x):
        self.data.append(x)

    def pop(self):
        self.data.pop()

    def top(self):
        return self.data[-1]


n = int(input())

st = Stack()

for _ in range(n):

    cmd = input().split()

    if cmd[0] == "PUSH":
        st.push(int(cmd[1]))

    elif cmd[0] == "POP":
        st.pop()

    else:
        print(st.top())
```
